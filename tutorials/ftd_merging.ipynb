```python
import pandas as pd

def load_transpose_and_merge_on_DID(csv_to_transpose, second_csv):
    """
    Transpose the first CSV on DID, clean the second CSV by dropping
    columns ending with |CSF or |PLASMA, then merge on DID.
    """
    # --- First CSV ---
    df = pd.read_csv(csv_to_transpose, index_col=0)
    df.index = df.index.astype(str) + "|CSF"
    df_t = df.transpose().reset_index().rename(columns={"index": "DID"})

    # Cast DID to int if possible
    try:
        df_t["DID"] = df_t["DID"].astype(int)
    except ValueError:
        df_t["DID"] = df_t["DID"].astype(str)

    # --- Second CSV ---
    second_df = pd.read_csv(second_csv)

    # Drop columns ending with |CSF or |PLASMA
    drop_cols = [c for c in second_df.columns if c.endswith("|CSF") or c.endswith("|PLASMA")]
    if drop_cols:
        second_df = second_df.drop(columns=drop_cols)

    print(second_df.columns)
    # Cast DID to same type as df_t
    if df_t["DID"].dtype == int:
        second_df["DID"] = second_df["DID"].astype(int)
    else:
        second_df["DID"] = second_df["DID"].astype(str)

    # --- Merge ---
    merged = pd.merge(df_t, second_df, on="DID", how="left")
    print(merged.shape)

    out_path = csv_to_transpose.replace(".csv", "_merged.csv")
    merged.to_csv(out_path, index=False)
    print(f"Merged shape: {merged.shape}")
    print(f"Saved to: {out_path}")
    return merged
```

```python
import pandas as pd

def load_transpose_and_merge_on_label_sporadic(csv_to_transpose, second_csv, did_colnames_csv=None):
    """
    Transpose the first CSV on DID, clean the second CSV by dropping
    columns ending with |CSF or |PLASMA, then merge on DID.
    Also, recode and rename additional columns in the second CSV:
    - 'sex' column: 1 -> "M", 2 -> "F", and rename to SEX_AT_BIRTH
    - 'age' column: rename to AGE_AT_VISIT
    
    If did_colnames_csv is provided, filter to only keep DIDs that match the column names.
    """
    # --- Read DIDs to keep from CSV if provided ---
    dids_to_keep = None
    if did_colnames_csv is not None:
        df_did = pd.read_excel(did_colnames_csv, nrows=0)
        dids_to_keep = {str(x) for x in df_did.columns if x not in ["", "DID", "index"]}
    
    # --- First CSV ---
    df = pd.read_csv(csv_to_transpose, index_col=0)
    df.index = df.index.astype(str) + "|CSF"
    df_t = df.transpose().reset_index().rename(columns={"index": "DID"})
    print(df_t.shape)

    # Cast DID to int if possible
    try:
        df_t["DID"] = df_t["DID"].astype(int)
        did_dtype = int
    except ValueError:
        df_t["DID"] = df_t["DID"].astype(str)
        did_dtype = str
    
    # Filter by allowed DIDs if needed
    if dids_to_keep is not None:
        if did_dtype == int:
            dids_to_keep_cast = {int(d) for d in dids_to_keep if d.isdigit()}
        else:
            dids_to_keep_cast = dids_to_keep
        before_count = df_t.shape[0]
        df_t = df_t[df_t["DID"].isin(dids_to_keep_cast)]
        print(f"Filtered DIDs: {before_count} -> {df_t.shape[0]}")

    # --- Second CSV ---
    second_df = pd.read_csv(second_csv)

    # Rename specimen_barcode (if present) to DID
    if "specimen_barcode" in second_df.columns:
        second_df = second_df.rename(columns={"specimen_barcode": "DID"})

    # Drop columns ending with |CSF or |PLASMA
    drop_cols = [c for c in second_df.columns if c.endswith("|CSF") or c.endswith("|PLASMA")]
    if drop_cols:
        second_df = second_df.drop(columns=drop_cols)

    # Recode 'sex' column, then rename to SEX_AT_BIRTH
    if "sex" in second_df.columns:
        second_df["sex"] = second_df["sex"].replace({1: "M", 2: "F"})
        second_df = second_df.rename(columns={"sex": "SEX_AT_BIRTH"})
    # Rename 'age' column, if present, to AGE_AT_VISIT
    if "age" in second_df.columns:
        second_df = second_df.rename(columns={"age": "AGE_AT_VISIT"})

    print(second_df.columns)
    # Cast DID to same type as df_t
    if did_dtype == int:
        second_df["DID"] = second_df["DID"].astype(int)
    else:
        second_df["DID"] = second_df["DID"].astype(str)

    # --- Merge ---
    merged = pd.merge(df_t, second_df, on="DID", how="left")
    print(merged.shape)

    out_path = csv_to_transpose.replace(".csv", "_merged.csv")
    merged.to_csv(out_path, index=False)
    print(f"Merged shape: {merged.shape}")
    print(f"Saved to: {out_path}")
    return merged
```

```python
csv_to_transpose = "/scratch/lcornelis/data/data_louisa/SporadicFTD/raw/cleanDat.Soma.CSFneat.macwide_allftdannotations.csv"
second_csv = "/scratch/lcornelis/data/data_louisa/SporadicFTD/raw/MAC wide traits.ctl and ftd.csv"
col_names = "/scratch/lcornelis/data/data_louisa/SporadicFTD/raw/cleanDat.Soma.CSFneat.5SDWinsor.146sporadicFTD.xlsx"
load_transpose_and_merge_on_label_sporadic(csv_to_transpose, second_csv, col_names)
```
