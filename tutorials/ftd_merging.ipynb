```python
import pandas as pd

def load_transpose_and_merge_on_DID(csv_to_transpose, second_csv):
    """
    Transpose the first CSV on DID, clean the second CSV by dropping
    columns ending with |CSF or |PLASMA, then merge on DID.
    """
    # --- First CSV ---
    df = pd.read_csv(csv_to_transpose, index_col=0)
    df.index = df.index.astype(str) + "|CSF"
    df_t = df.transpose().reset_index().rename(columns={"index": "DID"})

    # Cast DID to int if possible
    try:
        df_t["DID"] = df_t["DID"].astype(int)
    except ValueError:
        df_t["DID"] = df_t["DID"].astype(str)

    # --- Second CSV ---
    second_df = pd.read_csv(second_csv)

    # Drop columns ending with |CSF or |PLASMA
    drop_cols = [c for c in second_df.columns if c.endswith("|CSF") or c.endswith("|PLASMA")]
    if drop_cols:
        second_df = second_df.drop(columns=drop_cols)

    print(second_df.columns)
    # Cast DID to same type as df_t
    if df_t["DID"].dtype == int:
        second_df["DID"] = second_df["DID"].astype(int)
    else:
        second_df["DID"] = second_df["DID"].astype(str)

    # --- Merge ---
    merged = pd.merge(df_t, second_df, on="DID", how="left")
    print(merged.shape)

    out_path = csv_to_transpose.replace(".csv", "_merged.csv")
    merged.to_csv(out_path, index=False)
    print(f"Merged shape: {merged.shape}")
    print(f"Saved to: {out_path}")
    return merged
```

```python
second_csv = "/scratch/lcornelis/data/data_louisa/FTD/raw/ALLFTD_dataset_for_nina_louisa_071124_age_adjusted.csv"
csv_to_transpose = "/scratch/lcornelis/data/data_louisa/FTD/raw/cleanDat.Soma.CSFneat.5SDWinsor.csv"
load_transpose_and_merge_on_DID(csv_to_transpose, second_csv)
```
